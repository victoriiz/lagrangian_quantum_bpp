```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
sizes = [10,20,40,50,120]

lagrange_times = [0.0063, 0.014, 0.054, 0.092, 0.57]
base_times = [0.005, 0.015, 0.087, 0.199, 1.155]

optimal_sol = [6,10,19,23,49]
base_sol = [6,10,20,25,57]
lagrange_sol = [6,10,19,24,50]
```

```python
#Plot the times across the problem sizes
plt.figure(figsize=(12, 5))

plt.plot(sizes, lagrange_times, label='Lagrange Hybrid Model', marker='o', color = 'cornflowerblue')
plt.plot(sizes, base_times, label='Baseline Hybrid Model', marker='o', color = 'goldenrod')

plt.xlabel('Problem Size (items)')
plt.ylabel('Time (seconds)')
plt.title('CQM Construction Time Comparison Across Problem Sizes')
plt.legend()
plt.show()
```

```python
#Plot and compare the bin solutions
bar_width = 0.25
index = np.arange(len(sizes))

plt.bar(index, optimal_sol, width=bar_width, label='Known Optimal Solution', color='lightcoral',align='center')
plt.bar(index + bar_width, lagrange_sol, width=bar_width, label='Lagrangian Relaxation Hybrid Model', color='cornflowerblue',align='center')
plt.bar(index + 2*bar_width, base_sol, width=bar_width, label='Original Hybrid Model', color='palegoldenrod',align='center')

plt.xlabel('Problem Sizes (items)')
plt.ylabel('Number of Bins in Solution')
plt.title('Bin Solution Comparison Across Problem Sizes')
plt.xticks(index + bar_width / 3, sizes)
plt.legend()
```
